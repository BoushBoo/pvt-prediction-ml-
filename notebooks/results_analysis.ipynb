# Analyze saved evaluation predictions
This notebook never fits models. Metrics and plots are reconstructed from the same saved outer-fold predictions. Fold standard deviations are descriptive, not confidence intervals.


```python
from pathlib import Path
import json
import pandas as pd
from pvt_ml.evaluation import summarize
from pvt_ml.plotting import plot_predictions
RUN = Path("../results/PVT_1225_Pb_seed42")
metadata = json.loads((RUN / "metadata.json").read_text())
assert metadata["status"] == "complete", "Run is incomplete; inspect metadata"
predictions = pd.read_csv(RUN / "predictions.csv")
fold_metrics, summary, pooled = summarize(predictions)
summary
```
