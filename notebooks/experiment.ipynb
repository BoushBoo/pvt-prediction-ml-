# Configurable PVT experiment
Install from the repository root: `pip install -e ".[boosting,neural,notebook]"`. Choose a local data file and a fresh output directory. Full nested searches can take hours; use an explicit model list for a small run. No observations are cleaned or removed.


```python
from pathlib import Path
import hashlib
from pvt_ml.data import load_data
from pvt_ml.models import PHASE1, PHASE2
from pvt_ml.experiment import run_experiment
```

```python
# Paths relative to this notebook (or use absolute paths).
DATA = Path("../Dataset/PVT_1225.xls")
TARGET = "Pb"  # Pb or Bob
MODELS = PHASE1 + PHASE2  # Or e.g. ("linear", "extra_trees")
OUTPUT = Path("../results/PVT_1225_Pb_seed42")
OUTER_FOLDS = 5
INNER_FOLDS = 5
SEED = 42
SEARCH_SPACES = None  # Explicit per-model overrides; recorded in metadata
```

```python
df = load_data(DATA)
source = {"filename": DATA.name, "file_sha256": hashlib.sha256(DATA.read_bytes()).hexdigest()}
summary = run_experiment(df, TARGET, OUTPUT, MODELS, OUTER_FOLDS, INNER_FOLDS, SEED, search_spaces=SEARCH_SPACES, source=source)
summary
```

Outer test folds never select parameters or stopping epochs. Each model family has one training-selected estimator per outer fold. Results describe new samples from the same mixture, not unseen wells. Comparing model families does not give an unbiased estimate of a subsequently chosen overall winner.
